```python
from abc import ABC, abstractmethod
```

```python
class Ele_con(ABC):
    @abstractmethod
    def start(self):
        pass
```

```python
class Tv(Ele_con):
    def start(self):
        print("TV")
class MicroWave(Ele_con):
    def start(self):
        print("MV")
class Fan(Ele_con):
    def start(self):
        print("Fan")
```

```python
ele = [Tv(), MicroWave(), Fan()]
```

```python
for e in ele:
    e.start()
```
