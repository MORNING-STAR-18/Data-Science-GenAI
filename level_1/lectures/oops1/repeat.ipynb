```python
class OurBankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
    def withdraw(self, amount):
        if amount > self.balance:
            print("insufficent amount")
        else:
            self.balance -= amount
        print(self.name, " withdraw :", amount, ". New Balance :", self.balance)
```

```python
user_1 = OurBankAccount("Morning Star", 1000)
```

```python
user_1.withdraw(100)
```

```python
class Robot:
    def __init__(self, customer_name = None):
        print("hello, welcome to robot")
        self.name = "chitti"
        self.ram_in_gb = 10
        self.version = "0.0.1"
        self.customer_name = customer_name
        print(customer_name)
    def config(self):
        print("Ram :", self.ram_in_gb)
        print("version:", self.version)
        print("robot name :", self.name)
    def customer_details(self):
        print("Costmer Name :", self.customer_name)
    
```

```python
r1 = Robot("Morning Star")
```

```python
r1.config()
```

```python
r1.customer_details()
```

```python
r2 = Robot()
```

```python
r2.customer_details()
```

```python
r2.config()
```

```python
class batch:
    def __init__(self):
        self.students = []
    def add_students(self, *students):
        for student in students:
            self.students.append(student)
    def show_name(self):
        for i in self.students:
            print(i[0])
```

```python
ds_batch = batch()
```

```python
ds_batch.add_students(["morning star", 10])
ds_batch.add_students(["iong", 123], ["ioehgiofdng", 453], ["eiohggioengikorhg", 457890])
```

```python
ds_batch.show_name()
```
