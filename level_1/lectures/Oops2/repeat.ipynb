```python
class MotherBoard:
    def __init__(self, ram, cpu_speed, cpu_type):
        self.ram = ram
        self.cpu_speed = cpu_speed
        self.cpu_type = cpu_type
    def Start(self):
        print("power flow started")
    def stop(self):
        print("power flow stopped")
```

```python
class Computer(MotherBoard):
    def __init__(self, ram, cpu_speed, cpu_type):
        super().__init__(ram, cpu_speed, cpu_type)
        # self.os = self.cpu_type
        self.os = self.__check_os_and_set_os_type()
    def __check_os_and_set_os_type(self):
        if self.cpu_type == "mac-chip":
            return "mac"
        return "other"
    def check_my_os(self):
        print("user has :", self.os, "os")
```

```python
cmp1 = Computer("12gb", "4.4ghz", "mac-chip")
```

```python
cmp1._Computer__check_os_and_set_os_type()
```

```python
cmp1.check_my_os()
```

```python
cmp1.cpu_type = "dsoifn"
```

```python
cmp1.check_my_os()
```

```python
from abc import ABC, abstractmethod
```

```python
class Payment(ABC):
    @abstractmethod
    def pay(self, amount, currency):
        pass
    @abstractmethod
    def refund(self):
        pass
```

```python
class IPLDBank(Payment):
    def pay(self):
        print("Pay")
    def refund(self):
        print("refund")
    def new_user(self):
        print("hell0")
```

```python
obj_1 = IPLDBank()
```

```python
obj_1.refund()
```
