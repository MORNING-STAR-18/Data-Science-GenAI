```python
def join(operating_system, * args):
    final_output = ""
    for i in args:
        if operating_system == "windows" :
            final_output += i + "\\"
        elif operating_system == "linux" or operating_system == "mac":
            final_output += i + "/"
    final_output = final_output[:-1]
    return final_output
```

```python
print(join("linux", os.getcwd(), "temp", "dummy_2.txt"))
```
